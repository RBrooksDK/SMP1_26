```python
import numpy as np

rng = np.random.default_rng(seed=6)

# Simulate 100,000 mornings, each containing 50 orders.
orders = rng.choice(
    [1, 2, 6],
    size=(100_000, 50),
    p=[0.60, 0.30, 0.10]
)

# One total workload per morning.
totals = orders.sum(axis=1)

print(f"{'Mornings':>10} {'Average workload':>20} {'% within capacity':>20}")

for N in [100, 1_000, 10_000, 100_000]:
    mornings = totals[:N]

    # Calculation 1: average number of items per morning.
    average_workload = mornings.mean()

    # Calculation 2: fraction of mornings requiring at most 110 items.
    probability_of_fitting = (mornings <= 110).mean()

    print(
        f"{N:>10,}"
        f"{average_workload:>20.2f}"
        f"{100 * probability_of_fitting:>19.2f}%"
    )
```
